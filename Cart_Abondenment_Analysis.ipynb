### Data Cleaning & Preparation

```python
df_nov[df_nov['user_session'].isna()]
```

```python
# Remove missing values of user_sessions
df_nov = df_nov.dropna(subset = ['user_session'])
df_nov['user_session'].isna().sum()
```

```python
# Inspect duplicates
df_nov.duplicated().sum()
```

```python
df_nov[df_nov.duplicated(keep = False)].sort_values(by = ['user_session', 'event_time']).head(10)
```

```python
# Drop duplicates
df_nov = df_nov.drop_duplicates()
```

```python
# Convert dtype of event_time to datetime
df_nov['event_time'] = pd.to_datetime(df_nov['event_time'])
df_nov['event_time'].dtype
```

```python
# Inspect category_code column
df_nov['category_code'].dropna().str.split('.').str.len().value_counts().sort_index()
```

```python
# Inspect category_code's 4 parts rows unique values
df_nov[df_nov['category_code'].str.count(r'\.') == 3]['category_code'].nunique()
```

```python
## Create columns main_category, sub_category and product_type
parts = df_nov['category_code'].str.split('.')

df_nov['main_category'] = parts.str[0]
df_nov['sub_category'] = parts.where(parts.str.len() == 3).str[1]
df_nov['product_type'] = parts.str[-1]
```

```python
# Fill missing values with Unknown
df_nov.loc[df_nov['category_code'].isna(), 'category_code'] = 'Unknown'
df_nov.loc[df_nov['brand'].isna(), 'brand'] = 'Unknown'
df_nov.loc[df_nov['main_category'].isna(), 'main_category'] = 'Unknown'
df_nov.loc[df_nov['sub_category'].isna(), 'sub_category'] = 'Unknown'
df_nov.loc[df_nov['product_type'].isna(), 'product_type'] = 'Unknown'
```
